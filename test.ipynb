```python
from openai import OpenAI
import os
import pandas as pd
from dotenv import load_dotenv

# Load environment variables
load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# Load dataset
data = pd.read_csv("dataset/dataset_clean.csv")
context = " ".join(data.astype(str).values.flatten()[:1000])  # use first 1000 entries as context

# Define question
question = "中华人民共和国的法律框架是如何促进国家法治建设和发展的？"

# Create prompt
prompt = f"以下是一些资料：{context}\n\n问题：{question}\n请用中文回答。"

# Get response
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}]
)

# Display in universal format
answer = response.choices[0].message.content.strip()

print("========== Question ==========")
print(question)
print("========== Answer ============")
print(answer)
print("================================")
```
